```python
import pandas as pd

# Load the HorsePrices dataset
df = pd.read_csv("https://vincentarelbundock.github.io/Rdatasets/csv/Stat2Data/HorsePrices.csv")

# Clean up 
if 'rownames' in df.columns:
    df = df.drop(columns=['rownames'])

# Show dataset size using shape[0] and shape[1]
print("=== Dataset Overview ===")
print(f"Number of rows (horses):    {df.shape[0]}")
print(f"Number of columns:          {df.shape[1]}")
print("\n=== Statistics by Sex (m = male, f = female) ===\n")

for sex in ['m', 'f']:
    group = df[df['Sex'] == sex]
    
    print(f"\n--- Sex = {sex} ({len(group)} horses) ---")
    
    for col in ['Price', 'Height']:
        print(f"  {col} Statistics:")
        print(f"    Mean:     {group[col].mean():,.0f}")
        print(f"    Max:      {group[col].max():,.0f}")
        print(f"    Min:      {group[col].min():,.0f}")
        print(f"    Std:      {group[col].std():,.0f}")
        print(f"    Range:    {group[col].max() - group[col].min():,.0f}")
        print(f"    Median:   {group[col].median():,.0f}")
        print()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset from URL
url = "https://vincentarelbundock.github.io/Rdatasets/csv/Stat2Data/HorsePrices.csv"
df = pd.read_csv(url)

# Remove rownames column
if 'rownames' in df.columns:
    df = df.drop(columns=['rownames'])

# Show dataset size using shape[0] and shape[1]
print("=== Dataset Overview ===")
print(f"Number of rows (horses):    {df.shape[0]}")
print(f"Number of columns:          {df.shape[1]}")
print("\n Dataset loaded!\n")

# ==================== HISTOGRAMS BY SEX (for Age) ====================
for sex in ['m', 'f']:
    # Create a boolean flag (True/False series)
    if sex == 'm':
        is_male = df['Sex'] == 'm'      # This creates a boolean flag
        subset = df[is_male]            # Use the flag to filter
        title = 'Histogram of Age - Male Horses'
    else:
        is_female = df['Sex'] == 'f'
        subset = df[is_female]
        title = 'Histogram of Age - Female Horses'
  
    # Create histogram
    plt.figure(figsize=(9, 5))
    plt.hist(subset['Age'], bins=10, 
             color='skyblue' if sex == 'm' else 'lightcoral',
             edgecolor='black')
  
    plt.title(title, fontsize=15)
    plt.xlabel('Age (years)', fontsize=12)
    plt.ylabel('Number of Horses', fontsize=12)
    plt.grid(True, alpha=0.3)
    plt.show()
    
    print(f"→ Shown histogram for {sex.upper()} horses ({len(subset)} horses)")
```
