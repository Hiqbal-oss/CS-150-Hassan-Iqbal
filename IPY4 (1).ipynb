```python


print("Please input a positive integer")
num = int(input(": "))


if num <= 1:
    print(f"{num} is not prime.")
elif num == 2:
    print(f"{num} is prime.")
else:
    
    is_prime = True
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    
    if is_prime:
        print(f"{num} is prime.")
    else:
        print(f"{num} is not prime.")
```

```python


monthly_temps = {
    "january":   [43, 27],
    "february":  [48, 30],
    "march":     [58, 39],
    "april":     [70, 49],
    "may":       [79, 60],
    "june":      [87, 68],
    "july":      [89, 71],
    "august":    [88, 70],
    "september": [82, 63],
    "october":   [71, 51],
    "november":  [58, 40],
    "december":  [46, 31]
}



print("Louisville, KY Monthly Average Temperatures (°F)")
print("Enter a month name \n")

while True:
    user_input = input("Month: ").strip().lower()
    
    if user_input == "quit":
        print("Goodbye!")
        break
    
    # Check if the input matches a month key
    if user_input in monthly_temps:
        high, low = monthly_temps[user_input]
        print(f"Average temperatures for {user_input.capitalize()}:")
        print(f"  High: {high}°F")
        print(f"  Low:  {low}°F")
        print()
    else:
        print("Sorry, that's not a month.")
        print("Or type 'quit' to exit.\n")
```

```python

inventory = {
    'hammer':           [15, 12.49],
    'wrench':           [20, 7.99],
    'screwdriver':      [40, 3.99],
    'nailgun':          [10, 36.59],
    'padlock':          [25, 5.99],
    'roll_of_duct_tape':[30, 4.95]
}

def show_menu():
    print("\nHardware Store Inventory Menu")
    print("1. Look up an item (quantity and price)")
    print("2. Show items within a maximum budget")
    print("3. Show total quantity of all items in stock")
    print("4. Show total value of entire inventory")
    print("5. Quit")
    print("-----------------------------")

while True:
    show_menu()
    choice = input("Enter your choice (1-5): ").strip()
    
    if choice == "5" or choice.lower() == "quit":
        print("Thank you for using the inventory system. Goodbye!")
        break
    
    if choice == "1":
        item = input("Enter item name: ").strip().lower()
        if item in inventory:
            qty, price = inventory[item]
            print(f"{item.capitalize()}: {qty} available at ${price:.2f} each")
        else:
            print(f"Sorry, '{item}' is not in our inventory.")
    
    elif choice == "2":
        try:
            budget = float(input("Enter your maximum budget ($): "))
            if budget < 0:
                print("Budget cannot be negative.")
            else:
                affordable = []
                for item, data in inventory.items():
                    qty, price = data
                    if price <= budget:
                        affordable.append(item)
                
                if affordable:
                    print(f"Items you can afford (≤ ${budget:.2f}):")
                    for item in sorted(affordable):
                        price = inventory[item][1]
                        print(f"  - {item.capitalize()}: ${price:.2f}")
                else:
                    print("No items are available within your budget.")
        except ValueError:
            print("Please enter a valid number for the budget.")
    
    elif choice == "3":
        total_quantity = 0
        for qty, price in inventory.values():
            total_quantity += qty
        print(f"Total quantity of items in stock: {total_quantity}")
    
    elif choice == "4":
        total_value = 0.0
        # Using an explicit for loop as requested
        for item, data in inventory.items():
            qty, price = data
            total_value += qty * price
        print(f"Total value of entire inventory: ${total_value:.2f}")
    
    else:
        print("Invalid choice. Please enter a number between 1 and 5.")
```

```python

translations = {
    "hello": "hola",
    "goodbye": "adiós",
    "thank you": "gracias",
    "please": "por favor",
    "yes": "sí",
    "no": "no",
    "water": "agua",
    "food": "comida",
    "house": "casa",
    "friend": "amigo",
    "love": "amor",
    "sun": "sol",
    "book": "libro",
    "cat": "gato",
    "dog": "perro"
}

print("Enter a word to translate from English to Spanish (hit enter to stop)")

while True:
    word = input(": ").strip().lower()
    
    if word == "":
        print("Goodbye!")
        break
    
    if word in translations:
        # Show the translation in uppercase to match your sample style
        translation = translations[word].upper()
        print(f"The Spanish word for {word} is {translation}.")
    else:
        print(f"The Spanish word for {word} is not in this dictionary.")
    
    # Repeat the prompt
    print("\nEnter a word to translate from English to Spanish (hit enter to stop)")
```
