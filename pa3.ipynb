```python

side1 = float(input("Enter the length of side 1: "))
side2 = float(input("Enter the length of side 2: "))
side3 = float(input("Enter the length of side 3: "))

if side1 == side2 == side3:
    print("This is an equilateral triangle.")
elif side1 == side2 or side1 == side3 or side2 == side3:
    print("This is an isosceles triangle.")
else:
    print("This is a scalene triangle.")
```

```python

student_grades = [88, 72, 64, 91, 55]

class_average = sum(student_grades) / len(student_grades)
minimum_grade = min(student_grades)
maximum_grade = max(student_grades)

print("Class average:", class_average)
print("Minimum grade:", minimum_grade)
print("Maximum grade:", maximum_grade)

at_risk_count = 0
for grade in student_grades:
    if grade < 65:
        at_risk_count += 1

print("Number of students at risk:", at_risk_count)
```

```python
# A.
mult_of_2 = []
for num in range(1, 101):
    if num % 2 == 0:
        mult_of_2.append(num)

# B. 
mult_of_3 = []
for num in range(1, 101):
    if num % 3 == 0:
        mult_of_3.append(num)

# C.
mult_of_both = []
for num in mult_of_2:
    if num in mult_of_3:
        mult_of_both.append(num)

# D.
mult_of_either = []
for num in range(1, 101):
    if (num in mult_of_2) or (num in mult_of_3):
        mult_of_either.append(num)

print("Multiples of 2:", mult_of_2)
print("Multiples of 3:", mult_of_3)
print("Multiples of both:", mult_of_both)
print("Multiples of either:", mult_of_either)
```

```python
# A.
shopping_list = []

for i in range(7):
    item = input("Enter an item for your shopping list (or type 'done' to stop): ")
    if item.lower() == "done":
        break
    shopping_list.append(item)

# B
print("Total number of items:", len(shopping_list))

# C
print("List in reverse order:")
for i in range(len(shopping_list)-1, -1, -1):
    print(shopping_list[i])

# D
print("List in alphabetical order:")
sorted_list = sorted(shopping_list)
for item in sorted_list:
    print(item)

# E
duplicate_found = False

for i in range(len(shopping_list)):
    for j in range(i + 1, len(shopping_list)):
        if shopping_list[i] == shopping_list[j]:
            duplicate_found = True

if duplicate_found:
    print("At least one item appears more than once in your shopping list.")
else:
    print("No items are repeated in your shopping list.")
```
