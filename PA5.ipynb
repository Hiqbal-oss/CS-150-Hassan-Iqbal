```python

actSat = {"work", "school"}
actSun = {"work", "school", "birthday"}

print("Activities on Saturday:", actSat)
print("Activities on Sunday:", actSun)

both_days = actSat.intersection(actSun)
print("\nActivities done on BOTH days:", both_days)

either_day = actSat.union(actSun)
print("Activities done on AT LEAST one day:", either_day)
```

```python
def triangle_solve(leg1, leg2):
    
    hyp = (leg1**2 + leg2**2) ** 0.5

    
    perimeter = leg1 + leg2 + hyp

    
    area = 0.5 * leg1 * leg2

    
    return (hyp, perimeter, area)
```

```python
triangle_solve(3, 4)
```

```python
class Calculator:
    def __init__(self):
        self.prev_result = 0  
        self.total_calculations = 0

    def add(self, a, b):
        result = a + b
        self.prev_result = result
        self.total_calculations = self.total_calculations + 1  
        return result

    def subtract(self, a, b):
        result = a - b
        self.prev_result = result
        self.total_calculations = self.total_calculations + 1
        return result

    def multiply(self, a, b):
        result = a * b
        self.prev_result = result
        self.total_calculations = self.total_calculations + 1
        return result

    def divide(self, a, b):
        result = a / b  
        self.prev_result = result
        self.total_calculations = self.total_calculations + 1
        return result

    def power(self, a, b):
        result = a ** b
        self.prev_result = result
        self.total_calculations = self.total_calculations + 1
        return result

    def info(self):
        print("prev_result = " + str(self.prev_result))
        print("total_calculations = " + str(self.total_calculations))
```
